 - Ejercicio 1 : tarifa de estacionamiento

```python
prihora = 15.00
hora_add = 10.00
horas = int(input("Ingrese la cantidad de horas, en numero entero "))
if horas == 1:
    print(f"Total a pagar: {prihora}")
elif horas == 2:
    tot2 = (2 * prihora)
    print(f"Total a pagar: {tot2}")
elif horas >2:
    total =  (prihora * 2 + hora_add * horas)
    if total >= 200.00:
        print("Total a pagar: 200.00")
    if total <= 200.00:
        print(f"Total a pagar : {total}")
    
    

```

- Ejercicio 3 : Contador de digitos

```python
num = int(input("Dame un numero:"))
numoriginal = num
contador = 0
while num > 0:
    num = num//10
    contador = contador + 1
print (f"el numero {numoriginal} tiene {contador} digitos")
```

- Ejercicio 4 : Suma de dígitos

```python
num = int(input("Ingresa el numero que deseas sumar sus digitos:"))
numoriginal = num 
suma = 0
while num > 0:
    digito = num % 10
    suma = suma + digito
    num = num// 10
print("la suma de los digitos de", numoriginal, "es", suma)    
```
