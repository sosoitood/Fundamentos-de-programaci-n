##### Ejercicio 1: Area de un triangulo

```python
a = float(input("Ingresa la altura del triangulo en metros: "))
b = float(input("Ingresa la base del triangulo en metros: "))
area = (a * b) / 2
print("El area del triangulo es: ", area, "metros cuadrados")
```

##### Ejercicio 2: Conversión de celsius a fahrenheit

```python
celsius = float(input("Ingrese la temperatura en grados celsius: "))
f = celsius * 9/5 + 32
print(f"{celsius} grados celsius son igual a {f} grados fahrenheit")
```

##### Ejercicio 3: Interes simple

```python
cap = int(input("Ingrese el capital que desea que genere interes: "))
tasa = int(input("Ingrese la tasa de interes :"))
tiempo = int(input("Ingrese el tiempo que estara generando interes el capital:"))
interes = cap * (tasa/100) * tiempo
print("El interes simple es: ", interes)
```

##### Ejercicio 5

```python

x = input("Ingresa un número: ")
print("x es tipo", type(x).__name__)
try:
    resultado = x + 1
except TypeError:
    print("Error: no se puede sumar un str con un int")
x = int(x)
resultado = x + 1
print("Resultado correcto:", resultado)
```

##### Ejercicio 6: Conversión de horas a segundos 

```python

horas = float(input("Ingresa la cantidad de horas: "))
segundos = horas * 3600

print(f"{horas:g} horas equivalen a {segundos:g} segundos")
```

##### Ejercicio 7: Descuentos con strings

```python
precio = float(input("Ingresa el precio del artículo: $"))

descuento = precio * 0.15
precio_final = precio - descuento

print(f"Descuento (15%): ${descuento:.2f}")
print(f"Precio final: ${precio_final:.2f}")
```
