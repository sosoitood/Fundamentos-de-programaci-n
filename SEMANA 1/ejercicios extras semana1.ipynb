# ejercicio 1


```python
total_de_cuenta = float(input("Ingrese el total de la cuenta: "))
porcentaje_de_propina = float(input("Ingrese el porcentaje de propina: "))
numero_de_personas = float(input("Ingrese el número de personas: "))

propina = total_de_cuenta * (porcentaje_de_propina / 100)
total_con_propina = total_de_cuenta + propina
pago_por_persona = total_con_propina / numero_de_personas
 
print(f"Total de la cuenta: ${total_de_cuenta}")
print(f"Porcentaje de propina: {porcentaje_de_propina}% ")
print(f"Numero de personas: {numero_de_personas}" )

print(f"Propina: ${propina}")
print(f"Total con propina: ${total_con_propina}")
print(f"Pago por persona: ${pago_por_persona}")
```

# ejercicio 3



```python
calificacion_primer_parcial = int(input("Ingresa tu calificacion del primer parcial: "))
calificación_segundo_parcial = int(input("Ingresa tu calificación del segundo parcial: "))
calificación_tercer_parcial = int(input("Ingresa tu calificación del tercer parcial: "))
calificación_final = (calificacion_primer_parcial * 0.3 + calificación_segundo_parcial  * 0.3 + calificación_tercer_parcial * 0.4) 
print(f"Parcial 1 (30 %): {calificacion_primer_parcial}")
print(f"Parcial 2 (30 %): {calificación_segundo_parcial}")
print(f"Parcial 3 (40 %): {calificación_tercer_parcial}")
print(f"Tu calificación final: {calificación_final}") 
```

```python
pesos_mexicanos = int(input("ingrese la cantidad en pesos mexicanos: "))
tipo_de_cambio_USD = 17.57
tipo_de_cambio_EUR = 18.92
USD = pesos_mexicanos / tipo_de_cambio_USD
EUR = pesos_mexicanos / tipo_de_cambio_EUR

print(pesos_mexicanos, " MXN equivalen a:")
print("Cantidad en USD: ", USD)
print("Cantidad en EUR: ", EUR)
```
