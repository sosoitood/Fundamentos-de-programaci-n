```python
nombre_de_usuario = input("Ingresa su nombre: ")
horas_dedicadas_a_instagram = float(input("ingresa el numero de horas que dedicas a instagram: "))
horas_dedicadas_a_spotify = float(input("ingresa el numero de horas que dedicas a spotify: "))
horas_dedicadas_a_discord = float(input("ingresa el numero de horas dedicadas a discord: "))
horas_dedicadas_a_clash_royale = float(input("ingresa el numero de horas que dedicas a clash royale: "))
horas_dedicadas_a_whatsapp = float(input("ingresa el numero de horas dedicadas a whatsapp:"))

tiempo_total = (horas_dedicadas_a_instagram + horas_dedicadas_a_spotify + horas_dedicadas_a_discord + horas_dedicadas_a_clash_royale)

print(tiempo_total)
print(f"El usuario {nombre_de_usuario} dedica un total de {tiempo_total} horas a las redes sociales.")

porcentaje_del_tiempo_dedicado_al_uso_del_celular = (tiempo_total / 24) * 100
print(f"El usuario {nombre_de_usuario} dedica un total de {porcentaje_del_tiempo_dedicado_al_uso_del_celular:.2f}% de su tiempo al uso del celular.")

print(f"usuario: {nombre_de_usuario}")
print(f"tiempo total acumulado: {tiempo_total}")
print(f"porcentaje total calculado: {porcentaje_del_tiempo_dedicado_al_uso_del_celular} %")
```
