##### Ejercicio 7 

```python
num = int(input("Ingresa el numero del cual deseas guardar su tabla de multiplicar:"))
name = f"tabla_{num}.txt"
with open (f"tabla_{num}.txt", "w") as archivo:
    for i in range (1, 11):
        resultado = num * i
        archivo.write(f"{num} x {i} = {resultado}\n")
print(f"Tabla guardada en {name} ")       
```
