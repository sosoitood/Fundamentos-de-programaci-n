 # Ejercicio 1

```python
total = 0
visitante = 1
while True:
    costo = int(input(f"Precio del boleto {visitante}:"))
    if costo == 0:
        print("Menor de 3 años, sin cargo")
        visitante += 1
        continue
    total += costo
    if total >= 100:
        break 
    visitante += 1
    print(f"Total acumulado: $ {total}")
```
