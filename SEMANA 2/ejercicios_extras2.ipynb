```python
numero_visitantes = int(input("Ingrese el numero de visitantes: "))
suma_edades = 0
adultos = 0
for p in range(1, numero_visitantes + 1):
    edad_visitante = int(input(f"Ingrese la edad del visitante {p}"))
    suma_edades =  suma_edades + edad_visitante
    promedio = suma_edades / numero_visitantes
    if edad_visitante >= 18:
        adultos += 1
print("Adultos: ", adultos)
print("Promedio de edades:", promedio)

    

```
