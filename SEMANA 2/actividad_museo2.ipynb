# Actividad 2: Museo de Antropológia e Historia


```python
print("Bienvenido al museo de Antropologia e historia")

tabla_descuento = {
    "adulto_mayor": 0.12,
    "profesor":0.10,
    "estudiante":0.10,
    "ninguno":0.00
}
while True:
    cantidad = int(input("Ingrese el numero de visitantes"))
    if cantidad <= 0 :
         print("Error debe ingresar un numero mayor a 0")
         continue
    break 
total_general = 0
total_descuentos = 0
for visitante in range (1, cantidad + 1):
    print("Visitante:", visitante)

    while True:
        try:
            edad = int(input("Ingrese la edad del visitante:"))

            if edad <= 0:
                print("Error, la edad no puede ser menor a 0")
                continue
            break 
        except ValueError:
            print("Error. Ingrese una edad valida:")

    if edad < 3:
         precio_base = 0
         categoria = "Menor de 3 años"
    elif edad <= 17:
        precio_base = 89
        categoria = "Menor de edad"
    else:
        precio_base = 129
        categoria = "Mayor de edad"

    if edad < 3 :
        tipo = "Menor de 3 años"
        descuento = tabla_descuento["ninguno"]
        monto_descuento = 0
        precio_final = 0
        print("\nCategoria:", categoria)
        print("Tipo de visitante:", tipo)
        print("Entrada: SIN COSTO")
        print("Total a pagar: $0.00")

        continue
    
    elif edad < 21 :
        tipo = "Estudiante"
        descuento = tabla_descuento["estudiante"]

    elif edad >60 :
        tipo = "Adulto Mayor"
        descuento = tabla_descuento["adulto_mayor"]

    else:
        while True:
            if edad >= 3 and edad <= 60:
                print("\nTipo de visitante:")
                print("1. Profesor")
                print("2. Estudiante")
                print("3. Ninguno")
                opcion = input("Seleccione una opcion:")

                if opcion == "1":
                    tipo = "Profesor"
                    descuento = tabla_descuento["profesor"]

                    break

                elif opcion == "2":
                    tipo = "Estudiante"
                    descuento = tabla_descuento["estudiante"]

                    break

                elif opcion == "3":
                    tipo = "Boleto Normal"
                    descuento = tabla_descuento["ninguno"]

                    break

                else:
                    print("Opcion Invalida, ingrese 1. para profesor y 2. para estudiante:")
                    continue

    monto_descuento = precio_base * descuento
    precio_final = precio_base - monto_descuento
    total_general += precio_final
    total_descuentos += monto_descuento

    print("Resumen:")
    print("Edad:", edad)
    print("Categoria:", categoria)
    print("Tipo de visitante:", tipo)
    print(f"Descuento aplicado: {descuento * 100:.0f}%")
    print(f"Monto descontado: $ {monto_descuento:.2f}")
    print(f"Total a pagar: $ {precio_final:.2f}")

print("Total detallado:")
print("Numero de visitantes:", cantidad)
print(f"Total de descuentos: ${total_descuentos:.2f}")
print(f"TOTAL A PAGAR: ${total_general:.2f}")
print("Gracias por visitar el museo")
      
```
