Ejercicio 4: Clasificación por rango de edad con elif

```python
clasificacion_edad = int(input(" ¿Cuantos años tienes?"))
if clasificacion_edad <=3:
    print("Tu casificacion es: Bebe ")
elif clasificacion_edad <=12:
    print("Tu clasificacion es : Niño")
elif clasificacion_edad <=15:
    print("Tu clasificacion es : Adolescente")
```

 Ejercicio 5: Descuento por tipo de visitante

```python
tipo_de_visitante = input("Ingrese su tipo de visitante")
precio_boleto = 45
if tipo_de_visitante == "adulto_mayor":
    descuento_ad = precio_boleto * 0.12
    precio_ad = precio_boleto - descuento_ad
    print(" Usted es adulto mayor, el costo del boleto es: $ ", precio_ad)
elif tipo_de_visitante == "profesor":
    descuento_p= precio_boleto * 0.10
    precio_p = precio_boleto - descuento_p
    print(" Usted es profesor, el costo del boleto es:", precio_p)
elif tipo_de_visitante == "estudiante":
    descuento_e = precio_boleto * 0.10
    precio_e = precio_boleto - descuento_e
    print("Usted es estudiante, el costo del boleto es:", precio_e)
else:
    print("Sin descuento")


```

Ejercicio 11: suma progresiva

```python
num = int(input("Ingresa un numero positivo"))
contador = 1
suma = 0
while contador <= num:
    suma = suma + contador
    contador = contador + 1
print("La suma de 1 a 15 es:", suma)
```
